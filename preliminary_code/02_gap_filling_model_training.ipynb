# Feature engineering
The box below extracts the necessary features from the data (7 of them), normalizes each channel and then adds gaps to the data X, while at the same time it prserves the original data without the gaps as the label y.
- X in input = ['id','ts','species','temp','rh','vpd','swp','total_precip','rad','hour','month']
- y in input = ['GRO']
- X in output = ['GRO','temp','rh','vpd','swp','total_precip','rad','hour','month']
- y in output = ['GRO','temp','rh','vpd','swp','total_precip','rad','hour','month']

```python
def normalize(y_all, days_prediction, forecast):
    for i in range(len(y_all)):
        y_all[i] = (y_all[i] - np.average(y_all)) / np.std(y_all)
    return y_all

def rescale(x, y):
    for i in range(x.shape[0]):
        for j in range(x.shape[2]-1): #skip the last feature (month). If the segments are small, there will be one or two months and therefore only 0 and 1 when normalized
            if np.any(np.isnan(x[i,:,j])): # just in case for some reason an NaN passed through.
                print('ERROR!!!! There is a Nan in the data!!!!')
                print('Sample: ' + arg(i) + ', feature: ' + arg(j))
            mask = x[i,:,j] == -1 
            if len(mask[mask]) != 24*days_prediction: # This if statement is necessary because there could be features (such as temperature) that have -1 as a real value. However, it is unlikely that there will be many consecutive days with -1.
                mask[:] = False
            x_min = x[i,~mask,j].min()
            x_max = x[i,~mask,j].max()
            if np.abs(x_max - x_min) > 1e-4:
                x[i,~mask,j] = (x[i,~mask,j] - x_min) / (x_max - x_min)
                #y[i,:,j] = (y[i,:,j] - x_min) / (x_max - x_min)
                #y[i,:,j] = np.clip(y[i,:,j], 0, 1) #necessary to avoid bugs #TODO: this line is problematic
            else:
                x[i,~mask,j] = 0.5
                #y[i,:,j] = 0.5
            y_min = y[i,:,j].min()
            y_max = y[i,:,j].max()
            if np.abs(y_max - y_min) > 1e-4:
                y[i,:,j] = (y[i,:,j] - y_min) / (y_max - y_min)
            else:
                y[i,:,j] = 0.5
        x[i,:,-1] = x[i,:,-1]/12 # takes care of the last feature (months)
        y[i,:,-1] = y[i,:,-1]/12
    return x, y

def add_nans(m):
    for i in range(len(m)):
        if allFeatures:
            nb_missing_channels = np.random.randint(m.shape[2]-3) #The -3 makes sure that the features: 'SR','hour','month' are not considered
        else: nb_missing_channels = 0
        idx_missing_channels = np.sort(np.random.choice(range(1,m.shape[2]-2), size=nb_missing_channels, replace=False)) # The 1 ensures that 'SR' is not considered and the -2 that 'hour' and 'month' are not considered
        #gro_missing = np.random.randint(2) # 'SR' is considered separately
        gro_missing = 1 #this means that every sample will have a SR channel with a gap
        if gro_missing:
            idx_missing_channels = np.concatenate((np.array([0]), idx_missing_channels))
        for idx in idx_missing_channels:
            start_index = np.random.randint(days-days_prediction)*24//compression_factor
            end_index = start_index + days_prediction*24//compression_factor
            m[i, start_index:end_index, idx] = -1 # trick is to input missing values as -1
    return m


X_all = np.concatenate((y.reshape(y.shape[0],y.shape[1],1),X[:,:,[3,4,5,6,7,8,9,10]]),axis=2)
X_all = X_all.astype('float64')

y_all = X_all.copy() # [ML] The label or ground truth is the 7-channel time series (X_all in this case). Later, intervals of one or more X_all channels will be set to -1.

X_all, y_all = rescale(X_all, y_all) #need to rescale them both at the same time
X_all = add_nans(X_all)
#X_all, y_all = rescale(X_all, y_all) #need to rescale them both at the same time


print(X_all.shape)
print('')
print(y_all.shape)
```

# Train - validation split

```python
X_train = []
y_train = []
X_val = []
y_val = []
d_val_inds = {}

## ensure classes are split correctly between train and val
## + ensure same tree does not appear both in train and val
for label in np.unique(y_class): #separate first by label
    mask = y_class == label
    X_mask = X_all[mask]
    y_mask = y_all[mask]
    ids = X[mask][:,0,0]

    train_inds, val_inds = next(GroupShuffleSplit(test_size=.20, n_splits=2, random_state = rs).split(X_mask,y_mask,groups=ids))

    X_train.append(X_mask[train_inds])
    y_train.append(y_mask[train_inds])
    X_val.append(X_mask[val_inds])
    y_val.append(y_mask[val_inds])

    d_val_inds[label] = val_inds

X_train = np.concatenate(X_train, axis=0)
y_train = np.concatenate(y_train, axis=0)
X_val = np.concatenate(X_val,axis=0)
y_val = np.concatenate(y_val, axis=0)

## permutation
train_permutation = np.random.permutation(X_train.shape[0])
val_permutation = np.random.permutation(X_val.shape[0])
X_train = X_train[train_permutation]
y_train = y_train[train_permutation]
X_val = X_val[val_permutation]
y_val = y_val[val_permutation]
```

# Conclusion


*   the performance of the LSTM can probably be further improved by better hyper-parameter tuning
*   so far the framework only concerns separated chunks of N days. However can easily apply this method to a longer time-series by applying it sequentially on windows of N days
*   performance on weather / environment features is great because these time series can be the same in the train and the validation set
*   more generally, several trees can be geographically very close to each other, which leads to huge correlations in the data of these trees
